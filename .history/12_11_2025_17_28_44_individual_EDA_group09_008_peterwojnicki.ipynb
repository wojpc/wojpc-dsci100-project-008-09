### Summary of Initial Dataset Columns

#### Players Dataset

**experience:** "Amateur", "Beginner", "Regular", "Pro", "Veteran" assigned based on experience.

**subscribe:** "True" or "False" if subscribed (logical).

**hashedEmail:** Hashed Email of player.

**played_hours:** Number of hours played (dbl).

**name:** Name of player.

**gender:** "Female" or "Male" for gender of player.

**age:** Age of player (dbl).

#### Sessions Dataset

**hashedEmail:** Hashed Email of player.

**start_time:** Start time of session in format of DAY/MONTH/YEAR HOUR:MINUTE as chr.

**end_time:** End time of session in format of DAY/MONTH/YEAR HOUR:MINUTE as chr.

**original_start_time:** Start time recorded in UNIX time (milliseconds).

**original_end_time:** End time recorded in UNIX time (milliseconds).
